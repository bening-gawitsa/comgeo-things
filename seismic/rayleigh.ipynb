### Rayleigh Dispersion Equation

##### 1. Define the problem
1. **What is Rayleigh wave?**\
    Surface waves whose motion involves both **compressional (P-wave)** and **shear (S-wave)** components.
2. **Goals:**\
    Compute Rayleigh-wave **phase velocity $c$** at given frequency $\omega$ for a **layered elastic half-space**
3. **What parameters are causing the velocity become vary within layers?**
    - P-wave velocity of each layer ($\alpha_i$)
    - S-wave velocity of each layer ($\beta_i$)
    - Density of each layer ($\rho_i$)
    - Thickness of each layer ($d_i$)

##### 2. Physical Foundation
1. Medium: Elastic half-space
2. Assumptions:
    - harmonic time dependence $e^{i \omega t}$
    - horizontal wavenumber $k = \omega / v$
3. Governing equations:
    - Vertical propagation constants for P-waves
        $\begin{equation}
        r_{\alpha} = \sqrt{\frac{c^2}{\alpha^2} - 1}
        \end{equation}$
    - for S-waves
        $\begin{equation}
        r_{\beta} = \sqrt{\frac{c^2}{\beta^2} - 1}
        \end{equation}$
##### 3. The Layered Problem
1. The **difference properties** of each layers causing the continuity of velocity therefore the **displacements** and **stress** must be continuous too across each interface
$\begin{equation}
[u_z, u_x, \sigma_zz, \sigma_xx] = [u_z, u_x, \sigma_zz, \sigma_xx]_{i^+}
\end{equation}$
2. To adapt this behaviour we can apply Thomson-Haskell method to get 4x4 propagator matrix
$\begin{equation}
M_i(c,\omega) = [\zeta_i]
\end{equation}$
3. Trigonometric terms ($\zeta$)
$\begin{equation}
\cos(kr_a d)
\end{equation}$
$\begin{equation}
\sin(kr_a d)
\end{equation}$
$\begin{equation}
\cos(kr_b d)
\end{equation}$
$\begin{equation}
\sin(kr_b d)
\end{equation}$
4. **Rayleigh wave** is a **plane wave** with **vertical propagation**
